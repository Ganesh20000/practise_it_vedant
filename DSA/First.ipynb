## Two pointer

```python
arr = [1,2,3,4,5,5,5,5,6]

i = 0
for j in range(1, len(arr)):
    if arr[j] != arr[i]:
        i += 1
        arr[i] = arr[j]

print("Unique elements:", arr[:i+1])
print("Count:", i+1)
```

```python
arr = [1,2,3,4,5,5,5,5,6]


n=len(arr)
if n==1:
    print(1)

i=0
j=i+1
while j<n:
    if arr[j]!=arr[i]:
        i+=1
        arr[i],arr[j]=arr[j],arr[i]

    j+=1

# return i+1

print(i+1)
```

### left rotate array

```python
arr=[1,2,3,4,7,8]


temp=arr[0]

for i in range(1,len(arr)):
    arr[i-1]=arr[i]



arr[-1]=temp
print(arr)
```

```python
d=3
arr=[1,2,3,4,7,8,4,5,4,7,8,7,8,9,2]


temp=arr[:d]


for i in range(d,len(arr)):
    arr[i-d]=arr[i]


j=0

for i in range(len(arr)-d,len(arr)):
    arr[i]=temp[j]
    j+=1

print(arr)
```
