## Polymorphism

```python
class overloading:

    def __init__(self,a,b,*number):
        self.a=a
        self.b=b
        self.number=number



    def me(self):

        return self.a+self.b+sum(self.number)             
    


k=overloading(4,4,7,8,7)

print(k.me())

```

### Method overloading
Example Using *args

```python
class ti:

    def add(self,*arg):
        total=0

        for i in arg:
            total+=i


        return total
    

k=ti()


print(k.add(1,2,4))


print(k.add(7,8,9,85))
```

```python
class uy:

    def sum(self,a=None,b=None):
        if a!=None and b!=None:
            print(a+b)
        elif a!=None:
            print(a)
        else:
            print(None)



    
j=uy()

print(j.sum(1))


print(j.sum(74,7))
```
