## getter and setter

```python
## encapsulation
#^ Initialize the class 
class deg:
    # special method or magic method or dunder method
    # called as constructor
    def __init__(self):
        print(id(self))
        self.__name="ip"
        self.id=5
        self.age=78
        self.name='alice'
    #this is called method
    def second(self,score):
        print(f"i am a alice i score {score} ")


    def get_name(self):
        return self.__name
    

    def setter(self,names):
        self.__name=names

    def encap(self):
        self.__name="user not found"
    


d=deg()
print(d.second(50))

print(id(d))

print(d.get_name())

d.setter("elon ")
print(d.get_name())
```

## access element outside of class

```python
class deg:
    # special method or magic method or dunder method
    # called as constructor
    __user_id=1
    def __init__(self):
        print(id(self))
        self.__name="ip"
        self.id=5
        self.age=78
        self.name='alice'

        self.__user_id=deg.__user_id
        deg.__user_id+=1
    #this is called method
    def second(self,score):
        print(f"i am a alice i score {score} ")


    def get_name(self):
        return self.__name
    

    def setter(self,names):
        self.__name=names

    def encap(self):
        self.__name="user not found"
    
    @staticmethod
    def get():
        return deg.__user_id

    def set(n):
        deg.__user_id=n

d=deg()


d.get()

print(d.get())
d.set(25)

print(d.get())
```
