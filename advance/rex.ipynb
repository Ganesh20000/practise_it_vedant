```python
print("Regular expression")



import re
import warnings
warnings.filterwarnings('ignore')



string="A terror attack claimed by The Resistance Front (who later retracted the claim)[32] killed at least 26 civilians in Pahalgam, Jammu and Kashmir on 22 April 2025. India accused Pakistan of involvement, while Pakistan denied any such backing in 2025, Khwaja Muhammad Asif, Pakistan's Defence Minister in an interview with Sky News journalist Yalda Hakim, admitted that his country had supported terrorist groups for more than three decades, saying that Pakistan has been doing the dirty work of the western nations when asked about supporting and financing terrorism."


word='Pakistan'

# findall
m=re.findall(word,string)
print(m)


# match

string2='Pakistan is worst country'

m2=re.match(word,string)

print(m2)
# print(m2.group())


# search

# search it will look for the first occurence of regular expression
#sequence and deliver it 

M3=re.search(word,string)
print(M3)

print(M3.group())


# sub

x=re.sub("\s","@",string)
# print(x)


#split

x=re.split("",string)

# print(x)


print("*"*50)


# regular expression with meta character

text='Pakistan and state-sponsored terrorism refers to the involvement of Pakistan in terrorism 786 '

# find set of character between [a-m]

x=re.findall("[a-k]",text)
print(x)

# find all digit character

x=re.findall(r'\d\d\d',text)
print(x)



# find any character except newline character


txt='bobbbbzy the king'
x=re.findall("bob.y",txt)

print(x)


# start with 
x=re.findall("^b",txt)
print(x)

#end with

x=re.findall("king$",txt)
print(x)

# zero or more occurences


x=re.findall("^bob.*y",txt)
print(x)

# one ore more occurence

x=re.findall("^bob.+y",txt)
print(x)

txt='bobbbbzy the king'

# zero or one occurence

x=re.findall("ki.?g",txt)
print(x)




# exactly specified number of occurrences

t='Helllo bachho'

x=re.findall('^He.{3}o',txt)
print(x)



#either or

t="Rome was not build in a day "

x=re.findall("Rome|build",t)
print(x)



# capture and group 

t="ill never fall ill still"

y=re.findall('(ill|ll)',t)

print(y)


# special sequences

t="rome was not build in a  1-80 day_@"

#return a match where the string contain digit number 0-9
y=re.findall('\d-\d\d',t)
print(y)

# return a match where the string does not contaion digit 
y=re.findall("\D",txt)
print(y)

# return a match where the string contain a white space character

y=re.findall('\s',text)
print(y)


y=re.findall("\s",text)
print(y)
```
